```python
import PyPDF2
```

```python
# Extracting a specific page from a multipage pdf file

def extract_page(input_pdf, output_pdf, page_number):
    # Open the input PDF file
    with open(input_pdf, 'rb') as infile:
        reader = PyPDF2.PdfReader(infile)
        writer = PyPDF2.PdfWriter()

        # Check if the page number is valid
#        if page_number < 1 or page_number > reader.numPages:
#        raise ValueError(f"Invalid page number {page_number}. The document has {reader.numPages} pages.")

        # Get the specific page (page_number - 1 because page numbering starts from 0 in PyPDF2)
        page = reader.pages[page_number-1]
        writer.add_page(page)

        # Write the extracted page to the output PDF file
        with open(output_pdf, 'wb') as outfile:
            writer.write(outfile)

    print(f"Page {page_number} extracted to {output_pdf}")

# Example usage:
input_pdf = 'Mediclaim_policy.pdf'
output_pdf = 'Mediclaim_policy_80D.pdf'
page_number = 28  # Change this to the page number you want to extract

extract_page(input_pdf, output_pdf, page_number)
```
