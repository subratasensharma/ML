```python
import tkinter as tk
from tkinter import filedialog

def select_folder():
    folder_path = filedialog.askdirectory(title="Select folder with images")
    if folder_path:
        status_label.config(text=f"Selected folder:\n{folder_path}")
    else:
        status_label.config(text="No folder selected.")

# Create GUI
window = tk.Tk()
window.title("Folder Picker Test")
window.geometry("400x200")

title_label = tk.Label(window, text="Folder Selection Test", font=("Arial", 16))
title_label.pack(pady=20)

select_button = tk.Button(window, text="Select Folder", command=select_folder, height=2, width=20)
select_button.pack(pady=10)

status_label = tk.Label(window, text="Waiting for folder selection...", font=("Arial", 10))
status_label.pack(pady=10)

window.mainloop()
```
